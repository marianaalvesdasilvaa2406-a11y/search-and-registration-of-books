```python
class Livro:

    def __init__(self, titulo, autor, genero, quantidadeD):
        self.titulo = titulo
        self.autor = autor
        self.genero = genero
        self.quantidadeD = quantidadeD


class Biblioteca:

    def __init__(self):
        self.livros = []

    def adicionar_livro(self, livro):
        self.livros.append(livro)


biblioteca = Biblioteca()


# Livros iniciais

livro1 = Livro('O Hobbit', 'J.R.R. Tolkien', 'Fantasia', 5)
livro2 = Livro('Dom Casmurro', 'Machado de Assis', 'Romance', 3)
livro3 = Livro('Harry Potter', 'J.K. Rowling', 'Fantasia', 8)
livro4 = Livro('O Cortiço', 'Aluísio de Azevedo', 'Literatura clássica', 4)

biblioteca.adicionar_livro(livro1)
biblioteca.adicionar_livro(livro2)
biblioteca.adicionar_livro(livro3)
biblioteca.adicionar_livro(livro4)


# Cadastrar novo livro

def cadastrar_livro():

    titulo = input('Título: ').upper()
    autor = input('Autor: ').upper()
    genero = input('Gênero: ').upper()
    quantidadeD = int(input('Quantidade de livros no estoque: '))

    livro = Livro(titulo, autor, genero, quantidadeD)

    biblioteca.adicionar_livro(livro)

    print('Livro cadastrado com sucesso!')


# Listar livros

def listar_livros():

    for livro in biblioteca.livros:
        print('Título:', livro.titulo)
        print('Autor:', livro.autor)
        print('Gênero:', livro.genero)
        print('Quantidade:', livro.quantidadeD)
        print('------------------')


# Buscar livro

def buscar_livro():

    titulo_busca = input('Qual livro está buscando? ').upper()

    for livro in biblioteca.livros:

        if livro.titulo.upper() == titulo_busca:
            print('Livro encontrado!')
            print('Título:', livro.titulo)
            print('Autor:', livro.autor)
            print('Gênero:', livro.genero)
            print('Quantidade:', livro.quantidadeD)
            return

    print('Livro não encontrado!')


buscar_livro()
```

```python
# gráfico de gêneros

import matplotlib.pyplot as plt

def grafico_generos():

    generos = {}

    for livros in biblioteca.livros:

        if livros.genero in generos:
            generos[livros.genero] += livros.quantidadeD
        else:
            generos[livros.genero] = livros.quantidadeD

    plt.bar(generos.keys(), generos.values())

    plt.xlabel('Gênero')
    plt.ylabel('Quantidade')
    plt.title('Quantidade de livros por gênero')

    plt.show()


grafico_generos()
```
